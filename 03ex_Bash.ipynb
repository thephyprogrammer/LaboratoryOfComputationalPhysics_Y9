1\.c For each letter of the alphabet, count the number of students whose surname starts with that letter. 

```python
awk -F',' 'NR > 1 {print substr($1, 1, 1)}' LCP_copy.csv | sort | uniq -c | sort -nr | head -n 1
```

awk - text processing programming language (command line utility) - built specifically for text processing in csv files 
it can perform operations on text modify and process and generate reports and extract columns 
syntax: awk 'condition {action}' filename
$0 - entire line 
$1 etc - specific field 

NR = current record number (row)
NF = number fields in the current record (row)

BEGIN 'Main Block' END - execute instructions before and after processing a file 

-F',' -specifies delimiter 

sort - sorts alphabetically 
sort -nr - sorts numerically r signifies reverse order (largest to smallest)



```python
# same task using only bash 
max_count=0
max_letter=""

for letter in {A..Z}; do
    count=$(cut -d',' -f2 LCP_copy.csv | tail -n +2 | grep -ic "^$letter")
    echo "$letter: $count"

    if (( count > max_count )); then
        max_count=$count
        max_letter=$letter
    fi #ending statement for if statements
done

echo "Most common initial: $max_letter ($max_count students)"

```

Note that in bash during variable assignment there should be no spaces 
around the '=' sign. This will cause problems.

there is difference between "" and '' in bash. "$letter" would print the value of the variable while '$letter' passed to echo would print $letter.

1\.e Assume an obvious numbering of the students in the file (first line is 1, second line is 2, etc.), group students "modulo 18", i.e. 1,19,37,.. 2,20,38,.. etc. and put each group in a separate file  

```python
for group in {1..18}; do
    : >  "Group_${group}.csv"
done

i=0

while IFS= read -r student; do
    ((i++))
    group=$(( (i - 1) % 18 + 1 ))
    echo "$student" >> "Group_${group}.csv"
done < LCP_copy.csv

for group in {1..18}; do
    echo "Group $group: $(wc -l < Group_${group}.csv) students"
done
```

IFS= prevents trailing and leading whitespaces from being removed while reading 

$(( )) is a bash syntax which is called arithmetic expansion. It tells bash to evaluate the expression within the double parentheses and substitute the resulting number in variable. 

done < LCP_copy.csv directs the file into the loop that is written above this command. 

wc -1 counts the number of lines in each file.
